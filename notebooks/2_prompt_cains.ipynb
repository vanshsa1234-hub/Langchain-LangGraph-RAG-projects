```python
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
llm = ChatGroq(
    model="openai/gpt-oss-120b") 
out=StrOutputParser()
```

```python
prompts = ChatPromptTemplate.from_messages([
    ("system", "You are a translator and translate in {language}"),
    ("user", "{query}")
])
prompts
```

```python
final_prompts=prompts.format_messages(language="hindi",query="i love india")
final_prompts
```

```python
res=llm.invoke(final_prompts)
res.content
```

```python
prompts.invoke({"language":"hindi", "query":"i love you"})
```

```python
def transform_case(result:str):
    return result.upper()
```

```python
res=llm.invoke(prompts.invoke({"language":"hindi", "query":"i love you"}))
res.content
```

```python
chains=prompts | llm | out | transform_case
```

```python
res=chains.invoke({"language":"hingligh", "query":"i love langchain?"})
print(res)
```
