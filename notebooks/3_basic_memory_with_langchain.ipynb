```python
from langchain_groq import ChatGroq
llm=ChatGroq(model="openai/gpt-oss-120b")
```

```python
prompts=[
    {"role":"user","content":"Hello, my name is Vansh saini"},
    {"role":"ai", "content":'Hello, Vansh saini! how can i assist you'},
    {"role": "user","content":"what is my name?"}
]
res=llm.invoke(prompts)
res.content
```

## Manage History

##### here we can clearly seen that llm automatically finds itself "who is his"

```python
history=[]
while True:
    query=input("User: ")
    if query.lower() in {"exit","quit","bye"}:
        print("Good Bye 👋")
        break

    history.append({"role":"user","content":query})
    print("User: ",query)

    res=llm.invoke(history)
    history.append({"role":"ai","content":res.content})
    print("AI :",res.content,"\n")
```
