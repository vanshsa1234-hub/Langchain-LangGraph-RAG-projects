#### Solution is: Structured output

```python
from pydantic import BaseModel,Field
from typing import List ## very powerfull lib 
class ResponseStructure(BaseModel):
    name:str = Field(description="Complete Name")
    email:str = Field(description="Email Address")
    age:int= Field(description="Age")

structure_llm=llm.with_structured_output(ResponseStructure)
```

```python
res= structure_llm.invoke(f"Please give me only name, email and age from this text : {text}")
```

```python
class Movies(BaseModel):
    title:str = Field(description="Movie title")
    year:int = Field(description="Movie Released year.")

class AllMovies(BaseModel):
    movies:List[Movies]

movie_llm=llm.with_structured_output(AllMovies,method="json_schema")
```

```python
movie_llm.invoke("Give me one Sci-Fi Movie")
```

```python
res=movie_llm.invoke("Give me three Sci-Fi Movie")
```
