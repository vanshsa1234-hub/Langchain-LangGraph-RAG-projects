```python
from dotenv import load_dotenv
load_dotenv
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_number(a: int,b:int):
    """
        it will return the sum of two numbers

        Args:
            a: Number One
            b: Number Two
    """
    return a+b
```

```python
@tool
def multiply_number(a: int,b:int):
    """
        it will return the product of two numbers

        Args:
            a: Number One
            b: Number Two
    """
    return a*b
```

```python
add_number.invoke({"a":12,"b":3})
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent
```

```python
llm =ChatGroq(model="openai/gpt-oss-120b")
```

```python
agent=create_agent(
    model=llm,
    tools=[add_number, multiply_number],
    system_prompt="you are a math teacher,and always use tool for calculation."  
)
```

```python
response=agent.invoke({"messages":[{"role":"user","content":"what is 2 + 30 *10 ?"}]})
print(response["messages"][-1].content)
```
